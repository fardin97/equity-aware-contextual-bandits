# Equity-Aware Contextual Bandits for Alternative Delivery Recommendations

This notebook is an exploratory, simulation-based research pipeline for dynamic out-of-home (OOH) / alternative delivery location (ADL) recommendations using the 2024 NYC Citywide Mobility Survey (CMS).

## Scope

The supervised model estimates a broad revealed-behavior outcome (`ooh_any`). A transparent, assumed reward environment then maps each held-out person-day to six recommendation actions. Static policies and contextual bandits are evaluated inside that simulator.

This is **not** an offline reinforcement-learning study based on platform logs. CMS does not record recommendation offers, rejections, incentives, locker occupancy, or action-specific counterfactual outcomes. Results are simulator diagnostics, not causal or deployment evidence.

## Reproducibility and privacy

The public copy has no stored cell outputs or execution counts. It does not redistribute survey microdata or row-level predictions. Configure input and output locations with `CMS_DATA_DIR` and `CMS_OUTPUT_DIR`; see the repository README for expected files and setup.

## 1. Load and audit data

```python
# ============================================================
# 1. Load data
# ============================================================

for p in [TRIP_FILE, PERSON_FILE, DAY_FILE, HH_FILE]:
    if not p.exists():
        raise FileNotFoundError(p)

trip = pd.read_csv(TRIP_FILE, low_memory=False)
persons = pd.read_csv(PERSON_FILE, low_memory=False)
day = pd.read_csv(DAY_FILE, low_memory=False)
hh = pd.read_csv(HH_FILE, low_memory=False)

print("Trip:", trip.shape)
print("Persons:", persons.shape)
print("Day:", day.shape)
print("Household:", hh.shape)

try:
    dictionary = pd.read_excel(DICT_FILE, sheet_name="Sheet1", header=1)
    dictionary.columns = [str(c).strip() for c in dictionary.columns]
    print("Dictionary:", dictionary.shape)
except Exception as e:
    dictionary = None
    print("Dictionary not loaded:", e)

def parse_numeric(s):
    return pd.to_numeric(pd.Series(s).astype(str).str.replace(",", "", regex=False), errors="coerce")

def clean_numeric_codes(s, special_codes=SPECIAL_CODES):
    x = parse_numeric(s)
    return x.mask(x.isin(special_codes), np.nan)

def clean_binary_0_1(s):
    x = parse_numeric(s)
    return x.where(x.isin([0, 1]), np.nan)

def clean_categorical_codes(s, special_codes=SPECIAL_CODES):
    x = parse_numeric(s)
    out = x.round().astype("Int64").astype(str)
    out = out.replace({"<NA>": "Missing"})
    for code in special_codes:
        out = out.replace(str(code), f"Special_{code}")
    return out.astype("category")

def sigmoid(z):
    z = np.clip(z, -30, 30)
    return 1 / (1 + np.exp(-z))

def logit(p):
    p = np.clip(p, 1e-6, 1 - 1e-6)
    return np.log(p / (1 - p))

def minmax01(s):
    s = pd.Series(s, dtype="float")
    mn, mx = s.min(skipna=True), s.max(skipna=True)
    if pd.isna(mn) or pd.isna(mx) or mx == mn:
        return pd.Series(np.zeros(len(s)), index=s.index)
    return (s - mn) / (mx - mn)

def safe_auc(y, p, metric="roc"):
    y = np.asarray(y)
    p = np.asarray(p)
    if len(np.unique(y[~pd.isna(y)])) < 2:
        return np.nan
    try:
        if metric == "roc":
            return roc_auc_score(y, p)
        return average_precision_score(y, p)
    except Exception:
        return np.nan

def dict_lookup(col):
    if dictionary is None:
        return None
    hits = dictionary[dictionary["Column Name (system)"].astype(str).eq(col)]
    if hits.empty:
        return None
    return hits.iloc[0][["Dataset Name", "Column Name", "Column Description", "Expected/Allowed Values"]]

for c in ["delivery_5", "delivery_6", "delivery_7", "delivery_9", "shop_online", "packages"]:
    info = dict_lookup(c)
    if info is not None:
        print("\n" + c)
        print(info.to_string())
```

## 2. Construct package-delivery outcomes

Primary outcome for this notebook:

- `ooh_any = 1` if the respondent/day reports package delivery to work, another location such as a locker/pickup point, or another non-home location.

Strict locker/pickup sensitivity outcome:

- `adl_pickup = 1` if `delivery_7 == 1`, meaning package delivery to another location such as Amazon Locker, LockerNYC, or a package pickup point.

The broad OOH target is more consistent with the recommendation problem because the action set includes work/school-zone, errand-zone, transit-zone, and visited-zone non-home delivery recommendations. The strict `adl_pickup` target is still created and exported for sensitivity analysis.


```python
# ============================================================
# 2. Outcome construction
# ============================================================

delivery_cols = [c for c in day.columns if c.startswith("delivery_")]
for c in delivery_cols + ["shop_online"]:
    if c in day.columns:
        day[c + "_bin"] = clean_binary_0_1(day[c])

# Strict ADL/pickup-point/locker outcome.
day["adl_pickup"] = (day.get("delivery_7_bin", pd.Series(0, index=day.index)) == 1).astype(int)

# Broad out-of-home/non-home outcome. This is the main target in v3.
# delivery_6 = work, delivery_7 = another location / locker / pickup point,
# delivery_9 = other non-home/other location depending on CMS coding.
ooh_components = [c for c in ["delivery_6_bin", "delivery_7_bin", "delivery_9_bin"] if c in day.columns]
day["ooh_any"] = False
for c in ooh_components:
    day["ooh_any"] = day["ooh_any"] | (day[c] == 1)
day["ooh_any"] = day["ooh_any"].astype(int)

# Any reported package delivery, including home and OOH.
package_components = [c for c in ["delivery_5_bin", "delivery_6_bin", "delivery_7_bin", "delivery_9_bin"] if c in day.columns]
day["package_any"] = False
for c in package_components:
    day["package_any"] = day["package_any"] | (day[c] == 1)
day["package_any"] = day["package_any"].astype(int)

# Target-specific validity. For the broad OOH target, require all available OOH components to be valid.
validity_map = {
    "adl_pickup": ["delivery_7_bin"],
    "ooh_any": ooh_components,
    "package_any": package_components,
}
valid_cols = [c for c in validity_map.get(TARGET_NAME, ["delivery_7_bin"]) if c in day.columns]
if valid_cols:
    day["valid_target_response"] = day[valid_cols].notna().all(axis=1).astype(int)
else:
    day["valid_target_response"] = 1

# Keep the old name for backward compatibility in cells/exports.
day["valid_delivery_response"] = day["valid_target_response"]

target_col = TARGET_NAME
assert target_col in day.columns

rows = []
for c in ["adl_pickup", "ooh_any", "package_any", "shop_online_bin"]:
    if c in day.columns:
        y = day.loc[day["valid_target_response"] == 1, c]
        rows.append({
            "outcome": c,
            "n_valid_for_primary_target": int(y.notna().sum()),
            "positive_n": int((y == 1).sum()),
            "positive_rate": float((y == 1).mean())
        })
outcome_summary = pd.DataFrame(rows)
display(outcome_summary)
outcome_summary.to_csv(OUTPUT_DIR / "outcome_summary.csv", index=False)

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(outcome_summary["outcome"], outcome_summary["positive_rate"])
ax.set_ylabel("Positive rate")
ax.set_title(f"Outcome prevalence; primary target = {TARGET_NAME}")
ax.tick_params(axis="x", rotation=30)
fig.tight_layout()
fig.savefig(FIG_DIR / "outcome_prevalence.png", dpi=200)
plt.show()
```

## 3. Build linked-trip and activity-chain features

The trip file contains unlinked trip legs and linked trip IDs. The notebook collapses trip legs into linked trips, then aggregates linked trips into person-day activity chains.

```python
# ============================================================
# 3. Linked-trip and activity-chain construction
# ============================================================

PURPOSE_HOME = {1}
PURPOSE_WORK_SCHOOL = {2, 3, 4, 5}
PURPOSE_ERRAND_SHOP = {7, 8, 10}
PURPOSE_SOCIAL = {9}
PURPOSE_CHANGE_MODE = {11}

MODE_WALK = {1}
MODE_BIKE_SCOOTER = {2, 4}
MODE_AUTO = {8, 9}
MODE_TAXI_TNC = {5, 6}
MODE_TRANSIT = {12, 13}

CHAIN_CACHE_LINKED = OUTPUT_DIR / "linked_trips_2024.pkl"
CHAIN_CACHE_CHAIN = OUTPUT_DIR / "person_day_chain_features_2024.pkl"


def build_linked_trips(trip_df):
    needed = [
        "hh_id", "person_id", "day_id", "linked_trip_id", "linked_trip_num",
        "trip_id", "trip_num", "leg_num", "depart_hour", "depart_minute",
        "arrive_hour", "arrive_minute", "distance_miles", "duration_minutes",
        "o_cms_zone", "d_cms_zone", "mode_type", "mode_priority",
        "o_purpose_category", "d_purpose_category",
        "num_flags", "flag_distance", "flag_duration", "flag_mode",
        "trip_weight_citywide", "trip_weight_zonal", "n_legs"
    ]
    t = trip_df[[c for c in needed if c in trip_df.columns]].copy()

    for c in ["depart_hour", "depart_minute", "arrive_hour", "arrive_minute",
              "distance_miles", "duration_minutes", "o_cms_zone", "d_cms_zone",
              "mode_type", "mode_priority", "o_purpose_category", "d_purpose_category",
              "num_flags", "flag_distance", "flag_duration", "flag_mode", "n_legs"]:
        if c in t.columns:
            t[c] = clean_numeric_codes(t[c])

    t["depart_min_day"] = t["depart_hour"].fillna(0) * 60 + t["depart_minute"].fillna(0)
    t["arrive_min_day"] = t["arrive_hour"].fillna(0) * 60 + t["arrive_minute"].fillna(0)

    t["leg_walk"] = t["mode_type"].isin(MODE_WALK).astype(int)
    t["leg_bike_scooter"] = t["mode_type"].isin(MODE_BIKE_SCOOTER).astype(int)
    t["leg_auto"] = t["mode_type"].isin(MODE_AUTO).astype(int)
    t["leg_taxi_tnc"] = t["mode_type"].isin(MODE_TAXI_TNC).astype(int)
    t["leg_transit"] = t["mode_type"].isin(MODE_TRANSIT).astype(int)

    sort_cols = [c for c in ["hh_id", "person_id", "day_id", "linked_trip_num", "linked_trip_id", "leg_num", "trip_num"] if c in t.columns]
    t = t.sort_values(sort_cols)

    keys = ["hh_id", "person_id", "day_id", "linked_trip_id"]
    linked = t.groupby(keys, dropna=False).agg(
        linked_trip_num=("linked_trip_num", "first"),
        depart_min_day=("depart_min_day", "min"),
        arrive_min_day=("arrive_min_day", "max"),
        distance_miles=("distance_miles", "sum"),
        duration_minutes=("duration_minutes", "sum"),
        o_cms_zone=("o_cms_zone", "first"),
        d_cms_zone=("d_cms_zone", "last"),
        o_purpose_category=("o_purpose_category", "first"),
        d_purpose_category=("d_purpose_category", "last"),
        n_raw_legs=("trip_id", "count"),
        n_legs_reported=("n_legs", "max"),
        num_flags=("num_flags", "sum"),
        any_flag_distance=("flag_distance", "max"),
        any_flag_duration=("flag_duration", "max"),
        any_flag_mode=("flag_mode", "max"),
        has_walk=("leg_walk", "max"),
        has_bike_scooter=("leg_bike_scooter", "max"),
        has_auto=("leg_auto", "max"),
        has_taxi_tnc=("leg_taxi_tnc", "max"),
        has_transit=("leg_transit", "max"),
        trip_weight_citywide=("trip_weight_citywide", "first"),
        trip_weight_zonal=("trip_weight_zonal", "first")
    ).reset_index()

    linked["duration_from_clock"] = linked["arrive_min_day"] - linked["depart_min_day"]
    linked["duration_from_clock"] = linked["duration_from_clock"].where(linked["duration_from_clock"] >= 0, np.nan)

    linked["dest_home"] = linked["d_purpose_category"].isin(PURPOSE_HOME).astype(int)
    linked["dest_work_school"] = linked["d_purpose_category"].isin(PURPOSE_WORK_SCHOOL).astype(int)
    linked["dest_errand_shop"] = linked["d_purpose_category"].isin(PURPOSE_ERRAND_SHOP).astype(int)
    linked["dest_social"] = linked["d_purpose_category"].isin(PURPOSE_SOCIAL).astype(int)
    linked["dest_change_mode"] = linked["d_purpose_category"].isin(PURPOSE_CHANGE_MODE).astype(int)

    linked["is_multi_zone_linked_trip"] = (
        linked["o_cms_zone"].notna() & linked["d_cms_zone"].notna() &
        (linked["o_cms_zone"] != linked["d_cms_zone"])
    ).astype(int)
    return linked


def aggregate_zone_lists(zlong, keys, name):
    if zlong.empty:
        return pd.DataFrame(columns=keys + [name])
    out = zlong.groupby(keys, dropna=False)["zone"].agg(lambda s: sorted(set(s.dropna().astype(int)))).reset_index()
    out = out.rename(columns={"zone": name})
    return out


def make_zone_long(l):
    keys = ["hh_id", "person_id", "day_id"]
    origin_cols = keys + ["o_cms_zone", "o_purpose_category", "has_transit"]
    dest_cols = keys + ["d_cms_zone", "d_purpose_category", "has_transit"]
    o = l[origin_cols].rename(columns={"o_cms_zone": "zone", "o_purpose_category": "purpose"})
    d = l[dest_cols].rename(columns={"d_cms_zone": "zone", "d_purpose_category": "purpose"})
    z = pd.concat([o, d], ignore_index=True)
    z["zone"] = clean_numeric_codes(z["zone"])
    z["purpose"] = clean_numeric_codes(z["purpose"])
    z = z[z["zone"].notna() & ~z["zone"].isin(SPECIAL_CODES)].copy()
    z["zone"] = z["zone"].astype(int)
    return z


def build_person_day_chains(linked_df):
    l = linked_df.copy()
    keys = ["hh_id", "person_id", "day_id"]

    chain = l.groupby(keys, dropna=False).agg(
        linked_trip_count=("linked_trip_id", "nunique"),
        total_distance_miles=("distance_miles", "sum"),
        total_duration_minutes=("duration_minutes", "sum"),
        first_depart_min=("depart_min_day", "min"),
        last_arrive_min=("arrive_min_day", "max"),
        max_linked_trip_distance=("distance_miles", "max"),
        max_linked_trip_duration=("duration_minutes", "max"),
        total_raw_legs=("n_raw_legs", "sum"),
        total_num_flags=("num_flags", "sum"),
        any_flag_distance=("any_flag_distance", "max"),
        any_flag_duration=("any_flag_duration", "max"),
        any_flag_mode=("any_flag_mode", "max"),
        has_walk_day=("has_walk", "max"),
        has_bike_scooter_day=("has_bike_scooter", "max"),
        has_auto_day=("has_auto", "max"),
        has_taxi_tnc_day=("has_taxi_tnc", "max"),
        has_transit_day=("has_transit", "max"),
        work_school_trip_count=("dest_work_school", "sum"),
        errand_shop_trip_count=("dest_errand_shop", "sum"),
        social_trip_count=("dest_social", "sum"),
        change_mode_trip_count=("dest_change_mode", "sum"),
        home_dest_trip_count=("dest_home", "sum"),
        multi_zone_linked_trip_count=("is_multi_zone_linked_trip", "sum"),
    ).reset_index()

    chain["travel_span_minutes"] = chain["last_arrive_min"] - chain["first_depart_min"]
    chain["has_work_school_anchor"] = (chain["work_school_trip_count"] > 0).astype(int)
    chain["has_errand_shop_anchor"] = (chain["errand_shop_trip_count"] > 0).astype(int)
    chain["has_social_anchor"] = (chain["social_trip_count"] > 0).astype(int)
    chain["has_change_mode_anchor"] = (chain["change_mode_trip_count"] > 0).astype(int)
    chain["multi_zone_day"] = (chain["multi_zone_linked_trip_count"] > 0).astype(int)
    chain["multi_stop_day"] = (chain["linked_trip_count"] >= 3).astype(int)

    l_sorted = l.sort_values(["hh_id", "person_id", "day_id", "depart_min_day", "linked_trip_num"])
    first_rows = l_sorted.groupby(keys, dropna=False).first().reset_index()
    last_rows = l_sorted.groupby(keys, dropna=False).last().reset_index()
    chain = chain.merge(
        first_rows[keys + ["o_purpose_category", "o_cms_zone"]].rename(
            columns={"o_purpose_category": "first_origin_purpose", "o_cms_zone": "first_origin_zone"}
        ),
        on=keys, how="left"
    )
    chain = chain.merge(
        last_rows[keys + ["d_purpose_category", "d_cms_zone"]].rename(
            columns={"d_purpose_category": "last_dest_purpose", "d_cms_zone": "last_dest_zone"}
        ),
        on=keys, how="left"
    )
    chain["starts_at_home_purpose"] = chain["first_origin_purpose"].isin(PURPOSE_HOME).astype(int)
    chain["ends_at_home_purpose"] = chain["last_dest_purpose"].isin(PURPOSE_HOME).astype(int)
    chain["complete_home_based_chain"] = ((chain["starts_at_home_purpose"] == 1) & (chain["ends_at_home_purpose"] == 1)).astype(int)

    z = make_zone_long(l)
    all_zones = aggregate_zone_lists(z, keys, "visited_zones")
    work_zones = aggregate_zone_lists(z[z["purpose"].isin(PURPOSE_WORK_SCHOOL)], keys, "work_school_zones")
    errand_zones = aggregate_zone_lists(z[z["purpose"].isin(PURPOSE_ERRAND_SHOP)], keys, "errand_shop_zones")
    social_zones = aggregate_zone_lists(z[z["purpose"].isin(PURPOSE_SOCIAL)], keys, "social_zones")
    transit_zones = aggregate_zone_lists(z[(z["has_transit"] == 1) | z["purpose"].isin(PURPOSE_CHANGE_MODE)], keys, "transit_compatible_zones")

    for zz in [all_zones, work_zones, errand_zones, social_zones, transit_zones]:
        chain = chain.merge(zz, on=keys, how="left")

    list_cols = ["visited_zones", "work_school_zones", "errand_shop_zones", "social_zones", "transit_compatible_zones"]
    for c in list_cols:
        chain[c] = chain[c].apply(lambda x: x if isinstance(x, list) else [])

    chain["n_unique_visited_zones"] = chain["visited_zones"].apply(len)
    chain["n_work_school_zones"] = chain["work_school_zones"].apply(len)
    chain["n_errand_shop_zones"] = chain["errand_shop_zones"].apply(len)
    chain["n_transit_compatible_zones"] = chain["transit_compatible_zones"].apply(len)
    return chain

# Build chain features. Reuse cache if available in the current output folder.
if CHAIN_CACHE_LINKED.exists() and CHAIN_CACHE_CHAIN.exists():
    linked = pd.read_pickle(CHAIN_CACHE_LINKED)
    chain = pd.read_pickle(CHAIN_CACHE_CHAIN)
else:
    linked = build_linked_trips(trip)
    chain = build_person_day_chains(linked)

# Clip implausible travel-day totals while preserving flags.
clip_specs = {
    "total_distance_miles": MAX_DAILY_DISTANCE_MILES,
    "total_duration_minutes": MAX_DAILY_DURATION_MINUTES,
    "travel_span_minutes": MAX_TRAVEL_SPAN_MINUTES,
    "max_linked_trip_distance": MAX_LINKED_TRIP_DISTANCE_MILES,
    "max_linked_trip_duration": MAX_LINKED_TRIP_DURATION_MINUTES,
}
for col, cap in clip_specs.items():
    if col in chain.columns:
        raw_col = col + "_raw"
        flag_col = col + "_outlier_flag"
        if raw_col not in chain.columns:
            chain[raw_col] = chain[col]
        chain[flag_col] = (chain[raw_col] > cap).astype(int)
        chain[col] = chain[raw_col].clip(upper=cap)

print("Linked trips:", linked.shape)
print("Person-day chains:", chain.shape)
print("Chain columns:", sorted(chain.columns))

linked.to_pickle(CHAIN_CACHE_LINKED)
chain.to_pickle(CHAIN_CACHE_CHAIN)

chain_summary = chain[[
    "linked_trip_count", "total_distance_miles", "total_duration_minutes",
    "n_unique_visited_zones", "has_work_school_anchor", "has_errand_shop_anchor",
    "has_transit_day", "multi_zone_day", "multi_stop_day", "complete_home_based_chain"
]].describe().T
display(chain_summary)
chain_summary.to_csv(OUTPUT_DIR / "chain_feature_summary.csv")
```

## 4A. Publication-grade external zone opportunity builder (optional)

The default notebook uses a **survey-derived proxy** so the bandit pipeline can run with only CMS files. For journal results, replace that proxy with external geospatial opportunity measures aggregated to CMS survey zones.

Recommended external opportunity layers:

- **CMS Survey Zones** from NYC Open Data: spatial polygons used to aggregate everything to `cms_zone`.
- **Existing parcel lockers** from OpenStreetMap: `amenity=parcel_locker`.
- **Candidate staffed pickup points** from OpenStreetMap/Overture: pharmacies, grocery/supermarkets, convenience stores, post offices, parcel/postal services.
- **Transit opportunity** from MTA GTFS `stops.txt` for subway and bus stop density.
- **Land-use intensity** from PLUTO/MapPLUTO if you want residential/commercial anchors.

Set `USE_LIVE_EXTERNAL_OPPORTUNITY = True` below to try live downloads. The code is defensive: if internet access or GIS dependencies are unavailable, it keeps using the survey proxy defined in the next section.

```python
# ============================================================
# 4A. Optional live external opportunity builder
# ============================================================

USE_LIVE_EXTERNAL_OPPORTUNITY = False  # Turn on for publication runs with internet + GIS packages.

CMS_ZONES_GEOJSON_URL = "https://data.cityofnewyork.us/resource/sx7h-iavd.geojson?$limit=5000"

# Current/static GTFS URLs. If any URL changes, download the zip manually and point GTFS_ZIP_FILES to local paths.
GTFS_ZIP_URLS = {
    "subway": "https://rrgtfsfeeds.s3.amazonaws.com/gtfs_supplemented.zip",
}
# Add versioned bus-feed ZIP files through CMS_GTFS_DIR. Legacy unencrypted
# endpoints from the working notebook are intentionally omitted here.

_gtfs_dir_override = os.environ.get("CMS_GTFS_DIR")
GTFS_ZIP_FILES = (
    {path.stem: path for path in Path(_gtfs_dir_override).expanduser().glob("*.zip")}
    if _gtfs_dir_override else {}
)

# OSM tags for existing lockers and candidate ADL/PUDO points.
# Keep parcel lockers separate from candidate pickup points because OSM locker coverage can be incomplete.
OSM_TAGS = {
    "amenity": ["parcel_locker", "post_office", "pharmacy"],
    "shop": ["supermarket", "grocery", "convenience", "department_store"],
    "office": ["courier", "logistics"],
}


def _require_geospatial_stack():
    try:
        import geopandas as gpd
        import shapely
        return gpd
    except Exception as e:
        raise ImportError(
            "External opportunity builder needs geopandas/shapely. "
            "Install with conda-forge or pip before running this cell."
        ) from e


def load_cms_zone_polygons(url=CMS_ZONES_GEOJSON_URL):
    """Load CMS survey-zone polygons from NYC Open Data."""
    gpd = _require_geospatial_stack()
    zones = gpd.read_file(url)
    # Socrata field names can vary slightly; normalize candidate zone fields.
    zone_candidates = [c for c in zones.columns if c.lower() in {"cms_zone", "survey_zone", "zone", "name"}]
    if "cms_zone" not in zones.columns:
        # In current NYC Open Data export, one of these fields should identify the survey zone.
        # Print columns to help users pick the correct one if needed.
        print("CMS zone polygon columns:", list(zones.columns))
        for cand in zone_candidates:
            vals = pd.to_numeric(zones[cand], errors="coerce")
            if vals.notna().sum() > 0:
                zones["cms_zone"] = vals.astype("Int64")
                break
    if "cms_zone" not in zones.columns:
        raise ValueError("Could not infer cms_zone field from CMS Survey Zones. Inspect printed columns.")
    zones["cms_zone"] = pd.to_numeric(zones["cms_zone"], errors="coerce").astype("Int64")
    zones = zones.dropna(subset=["cms_zone"]).copy()
    return zones[["cms_zone", "geometry"]].to_crs(4326)


def build_osm_opportunities_by_zone(cms_zones_gdf, tags=OSM_TAGS):
    """Pull OSM features inside the CMS-zone union and aggregate candidate ADL counts by zone."""
    gpd = _require_geospatial_stack()
    try:
        import osmnx as ox
    except Exception as e:
        raise ImportError("Install osmnx to pull OSM opportunity data live.") from e

    polygon = cms_zones_gdf.to_crs(4326).unary_union
    osm = ox.features_from_polygon(polygon, tags=tags)
    if osm.empty:
        return pd.DataFrame({
            "cms_zone": cms_zones_gdf["cms_zone"].astype(int),
            "locker_count_osm": 0,
            "post_office_count_osm": 0,
            "grocery_pharmacy_count_osm": 0,
            "candidate_pickup_poi_count_osm": 0,
        })

    osm = osm.reset_index()
    osm = gpd.GeoDataFrame(osm, geometry="geometry", crs=4326)
    # Convert polygons/lines to representative points before spatial join.
    osm["geometry"] = osm.geometry.apply(lambda g: g if g.geom_type == "Point" else g.representative_point())

    joined = gpd.sjoin(
        osm.to_crs(cms_zones_gdf.crs),
        cms_zones_gdf[["cms_zone", "geometry"]],
        how="inner",
        predicate="within",
    )

    def count_where(df, mask, name):
        return df.loc[mask].groupby("cms_zone").size().rename(name)

    amenity = joined.get("amenity", pd.Series(index=joined.index, dtype=object)).astype(str)
    shop = joined.get("shop", pd.Series(index=joined.index, dtype=object)).astype(str)
    office = joined.get("office", pd.Series(index=joined.index, dtype=object)).astype(str)

    out = pd.DataFrame({"cms_zone": cms_zones_gdf["cms_zone"].astype(int).unique()})
    counts = [
        count_where(joined, amenity.eq("parcel_locker"), "locker_count_osm"),
        count_where(joined, amenity.eq("post_office"), "post_office_count_osm"),
        count_where(joined, amenity.eq("pharmacy") | shop.isin(["supermarket", "grocery", "convenience", "department_store"]), "grocery_pharmacy_count_osm"),
        count_where(joined, amenity.isin(["parcel_locker", "post_office", "pharmacy"]) | shop.isin(["supermarket", "grocery", "convenience", "department_store"]) | office.isin(["courier", "logistics"]), "candidate_pickup_poi_count_osm"),
    ]
    for s in counts:
        out = out.merge(s.reset_index(), on="cms_zone", how="left")
    count_cols = [c for c in out.columns if c != "cms_zone"]
    out[count_cols] = out[count_cols].fillna(0)
    return out


def _read_gtfs_stops_from_zip(zip_path_or_url):
    import io, zipfile, requests
    if str(zip_path_or_url).startswith("http"):
        resp = requests.get(zip_path_or_url, timeout=60)
        resp.raise_for_status()
        zf = zipfile.ZipFile(io.BytesIO(resp.content))
    else:
        zf = zipfile.ZipFile(zip_path_or_url)
    with zf.open("stops.txt") as f:
        stops = pd.read_csv(f)
    return stops


def build_gtfs_stop_counts_by_zone(cms_zones_gdf, gtfs_urls=GTFS_ZIP_URLS, gtfs_files=GTFS_ZIP_FILES):
    """Aggregate MTA GTFS stop counts to CMS zones."""
    gpd = _require_geospatial_stack()
    all_stops = []
    sources = dict(gtfs_urls)
    sources.update(gtfs_files)
    for name, src in sources.items():
        try:
            stops = _read_gtfs_stops_from_zip(src)
            if {"stop_lat", "stop_lon"}.issubset(stops.columns):
                stops["gtfs_source"] = name
                all_stops.append(stops[["stop_id", "stop_name", "stop_lat", "stop_lon", "gtfs_source"]])
        except Exception as e:
            print(f"GTFS source failed: {name}: {e}")
    if not all_stops:
        return pd.DataFrame({"cms_zone": cms_zones_gdf["cms_zone"].astype(int), "transit_stop_count_gtfs": 0})
    stops = pd.concat(all_stops, ignore_index=True).drop_duplicates(subset=["stop_id", "stop_lat", "stop_lon", "gtfs_source"])
    stops_gdf = gpd.GeoDataFrame(
        stops,
        geometry=gpd.points_from_xy(stops["stop_lon"], stops["stop_lat"]),
        crs=4326,
    )
    joined = gpd.sjoin(stops_gdf.to_crs(cms_zones_gdf.crs), cms_zones_gdf[["cms_zone", "geometry"]], how="inner", predicate="within")
    out = joined.groupby("cms_zone").size().rename("transit_stop_count_gtfs").reset_index()
    base = pd.DataFrame({"cms_zone": cms_zones_gdf["cms_zone"].astype(int).unique()})
    out = base.merge(out, on="cms_zone", how="left").fillna({"transit_stop_count_gtfs": 0})
    return out


def build_real_zone_opportunity_table():
    """Build a publication-grade zone opportunity table from live external data."""
    zones = load_cms_zone_polygons()
    osm_counts = build_osm_opportunities_by_zone(zones)
    gtfs_counts = build_gtfs_stop_counts_by_zone(zones)

    opp = zones[["cms_zone"]].copy()
    opp["cms_zone"] = opp["cms_zone"].astype(int)
    opp = opp.merge(osm_counts, on="cms_zone", how="left").merge(gtfs_counts, on="cms_zone", how="left")
    num_cols = [c for c in opp.columns if c != "cms_zone"]
    opp[num_cols] = opp[num_cols].fillna(0)

    # Publication-safe composite score. Report components separately; use this only for policy simulation.
    score_terms = []
    weights = {
        "locker_count_osm": 0.30,
        "candidate_pickup_poi_count_osm": 0.30,
        "grocery_pharmacy_count_osm": 0.20,
        "post_office_count_osm": 0.10,
        "transit_stop_count_gtfs": 0.10,
    }
    for c, wgt in weights.items():
        if c in opp.columns:
            score_terms.append(wgt * minmax01(np.log1p(opp[c].fillna(0))))
    opp["adl_opportunity_score"] = minmax01(sum(score_terms)) if score_terms else 0.0
    opp["opportunity_source"] = "external_osm_gtfs"
    return opp


if USE_LIVE_EXTERNAL_OPPORTUNITY:
    try:
        external_zone_opp = build_real_zone_opportunity_table()
        external_path = OUTPUT_DIR / "zone_opportunity_external_osm_gtfs.csv"
        external_zone_opp.to_csv(external_path, index=False)
        ZONE_OPPORTUNITY_FILE = str(external_path)
        print("External zone opportunity file saved and activated:", ZONE_OPPORTUNITY_FILE)
        display(external_zone_opp)
    except Exception as e:
        print("Live external opportunity build failed. Falling back to survey proxy.")
        print("Reason:", repr(e))
```

## 4. Zone-level ADL opportunity features

The uploaded CMS data have CMS zones, not exact coordinates. For method development, this notebook creates a transparent survey-derived zone opportunity proxy. For publication, replace it with external geospatial data aggregated to CMS zones.

```python
# ============================================================
# 4. Zone opportunity table
# ============================================================

def build_proxy_zone_opportunity(hh_df, linked_df):
    h = hh_df.copy()
    h["home_cms_zone_num"] = clean_numeric_codes(h["home_cms_zone"]) if "home_cms_zone" in h.columns else np.nan

    home_counts = h.groupby("home_cms_zone_num").size().rename("hh_home_count").reset_index().rename(
        columns={"home_cms_zone_num": "cms_zone"}
    )

    l = linked_df.copy()
    for c in ["o_cms_zone", "d_cms_zone", "d_purpose_category"]:
        if c in l.columns:
            l[c] = clean_numeric_codes(l[c])

    dest_counts = l.groupby("d_cms_zone").size().rename("dest_trip_count").reset_index().rename(columns={"d_cms_zone": "cms_zone"})
    work_counts = l[l["d_purpose_category"].isin(PURPOSE_WORK_SCHOOL)].groupby("d_cms_zone").size().rename("work_school_dest_count").reset_index().rename(columns={"d_cms_zone": "cms_zone"})
    errand_counts = l[l["d_purpose_category"].isin(PURPOSE_ERRAND_SHOP)].groupby("d_cms_zone").size().rename("errand_shop_dest_count").reset_index().rename(columns={"d_cms_zone": "cms_zone"})
    social_counts = l[l["d_purpose_category"].isin(PURPOSE_SOCIAL)].groupby("d_cms_zone").size().rename("social_dest_count").reset_index().rename(columns={"d_cms_zone": "cms_zone"})
    transit_counts = l[l["has_transit"] == 1].groupby("d_cms_zone").size().rename("transit_trip_dest_count").reset_index().rename(columns={"d_cms_zone": "cms_zone"})

    zone_values = pd.concat([
        home_counts["cms_zone"], dest_counts["cms_zone"], work_counts["cms_zone"],
        errand_counts["cms_zone"], social_counts["cms_zone"], transit_counts["cms_zone"]
    ], ignore_index=True).dropna()
    all_zones = sorted(set(zone_values.astype(int).tolist()))

    opp = pd.DataFrame({"cms_zone": all_zones})
    for d in [home_counts, dest_counts, work_counts, errand_counts, social_counts, transit_counts]:
        opp = opp.merge(d, on="cms_zone", how="left")

    for c in opp.columns:
        if c != "cms_zone":
            opp[c] = opp[c].fillna(0)

    opp["residential_proxy"] = minmax01(np.log1p(opp["hh_home_count"]))
    opp["employment_proxy"] = minmax01(np.log1p(opp["work_school_dest_count"]))
    opp["errand_proxy"] = minmax01(np.log1p(opp["errand_shop_dest_count"]))
    opp["transit_proxy"] = minmax01(np.log1p(opp["transit_trip_dest_count"]))
    opp["activity_density_proxy"] = minmax01(np.log1p(opp["dest_trip_count"]))

    opp["adl_opportunity_score"] = (
        0.20 * opp["residential_proxy"] +
        0.20 * opp["employment_proxy"] +
        0.30 * opp["errand_proxy"] +
        0.20 * opp["transit_proxy"] +
        0.10 * opp["activity_density_proxy"]
    )
    opp["adl_opportunity_score"] = minmax01(opp["adl_opportunity_score"])
    return opp

if ZONE_OPPORTUNITY_FILE is not None and Path(ZONE_OPPORTUNITY_FILE).exists():
    zone_opp = pd.read_csv(ZONE_OPPORTUNITY_FILE)
    if "cms_zone" not in zone_opp.columns:
        raise ValueError("External opportunity file must contain cms_zone.")
    num_cols = [c for c in zone_opp.columns if c != "cms_zone" and pd.api.types.is_numeric_dtype(zone_opp[c])]
    if "adl_opportunity_score" not in zone_opp.columns:
        if len(num_cols) == 0:
            raise ValueError("External opportunity file needs numeric opportunity columns.")
        score = sum(minmax01(np.log1p(zone_opp[c].fillna(0))) for c in num_cols) / len(num_cols)
        zone_opp["adl_opportunity_score"] = minmax01(score)
    zone_opp["opportunity_source"] = "external"
else:
    zone_opp = build_proxy_zone_opportunity(hh, linked)
    zone_opp["opportunity_source"] = "survey_proxy"

zone_opp["cms_zone"] = clean_numeric_codes(zone_opp["cms_zone"]).astype("Int64")
display(zone_opp)
zone_opp.to_csv(OUTPUT_DIR / "zone_opportunity_table.csv", index=False)

fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(zone_opp["cms_zone"].astype(str), zone_opp["adl_opportunity_score"])
ax.set_xlabel("CMS zone")
ax.set_ylabel("ADL opportunity score")
ax.set_title("Zone-level ADL opportunity score")
fig.tight_layout()
fig.savefig(FIG_DIR / "zone_opportunity_score.png", dpi=200)
plt.show()
```

```python
# ============================================================
# 5. Merge analytic table
# ============================================================

def prepare_household_features(hh_df):
    h = hh_df.copy()
    cols = [
        "hh_id", "home_cms_zone", "home_county", "residence_type", "residence_rent_own",
        "income_broad", "num_people", "num_adults", "num_kids", "num_workers",
        "num_students", "num_trips", "num_vehicles", "num_bicycles", "num_scooters",
        "hh_weight_citywide", "hh_weight_zonal"
    ]
    h = h[[c for c in cols if c in h.columns]].copy()

    numeric_cols = ["num_people", "num_adults", "num_kids", "num_workers", "num_students",
                    "num_trips", "num_vehicles", "num_bicycles", "num_scooters",
                    "hh_weight_citywide", "hh_weight_zonal"]
    for c in numeric_cols:
        if c in h.columns:
            h[c] = clean_numeric_codes(h[c])

    for c in ["home_cms_zone", "home_county", "residence_type", "residence_rent_own", "income_broad"]:
        if c in h.columns:
            h[c + "_cat"] = clean_categorical_codes(h[c])
            h[c + "_num"] = clean_numeric_codes(h[c])

    h["zero_vehicle_hh"] = (h.get("num_vehicles", pd.Series(np.nan, index=h.index)).fillna(999) == 0).astype(int)
    h["low_income_hh"] = h.get("income_broad_num", pd.Series(np.nan, index=h.index)).isin([1, 2]).astype(int)
    h["large_apartment_hh"] = h.get("residence_type_num", pd.Series(np.nan, index=h.index)).isin([4, 5, 6]).astype(int)
    return h

def prepare_person_features(person_df):
    p = person_df.copy()
    cols = [
        "hh_id", "person_id", "person_num", "age", "gender", "employment", "student",
        "work_in_region", "work_county", "work_cms_zone", "school_type",
        "school_in_region", "school_county", "school_cms_zone", "telework_freq",
        "commute_freq", "packages", "person_weight_citywide", "person_weight_zonal"
    ]
    p = p[[c for c in cols if c in p.columns]].copy()

    for c in ["age", "person_num", "person_weight_citywide", "person_weight_zonal"]:
        if c in p.columns:
            p[c] = clean_numeric_codes(p[c])

    for c in ["gender", "employment", "student", "work_in_region", "work_county", "work_cms_zone",
              "school_type", "school_in_region", "school_county", "school_cms_zone",
              "telework_freq", "commute_freq", "packages"]:
        if c in p.columns:
            p[c + "_cat"] = clean_categorical_codes(p[c])
            p[c + "_num"] = clean_numeric_codes(p[c])

    p["usual_package_alt_access"] = p.get("packages_num", pd.Series(np.nan, index=p.index)).isin([4, 5, 6, 7]).astype(int)
    p["usual_package_locker_access"] = p.get("packages_num", pd.Series(np.nan, index=p.index)).isin([6]).astype(int)
    return p

def prepare_day_features(day_df):
    d = day_df.copy()
    keep = [
        "hh_id", "person_id", "day_id", "day_num", "travel_date", "travel_dow",
        "is_complete", "num_trips", "shop_online", "telecommute_time",
        "day_weight_citywide", "day_weight_zonal",
        "adl_pickup", "ooh_any", "package_any", "valid_target_response", "valid_delivery_response"
    ] + [c + "_bin" for c in delivery_cols if c + "_bin" in d.columns]
    d = d[[c for c in keep if c in d.columns]].copy()

    for c in ["day_num", "travel_dow", "num_trips", "telecommute_time",
              "day_weight_citywide", "day_weight_zonal"]:
        if c in d.columns:
            d[c] = clean_numeric_codes(d[c])

    if "shop_online" in d.columns:
        d["shop_online_bin"] = clean_binary_0_1(d["shop_online"])
    d["weekend"] = d.get("travel_dow", pd.Series(np.nan, index=d.index)).isin([6, 7]).astype(int)
    return d

hh_feat = prepare_household_features(hh)
person_feat = prepare_person_features(persons)
day_feat = prepare_day_features(day)

analytic = day_feat.merge(person_feat, on=["hh_id", "person_id"], how="left")
analytic = analytic.merge(hh_feat, on="hh_id", how="left")
analytic = analytic.merge(chain, on=["hh_id", "person_id", "day_id"], how="left")

chain_numeric_fill0 = [
    "linked_trip_count", "total_distance_miles", "total_duration_minutes", "total_raw_legs",
    "total_num_flags", "has_walk_day", "has_bike_scooter_day", "has_auto_day",
    "has_taxi_tnc_day", "has_transit_day", "work_school_trip_count",
    "errand_shop_trip_count", "social_trip_count", "change_mode_trip_count",
    "home_dest_trip_count", "multi_zone_linked_trip_count", "travel_span_minutes",
    "has_work_school_anchor", "has_errand_shop_anchor", "has_social_anchor",
    "has_change_mode_anchor", "multi_zone_day", "multi_stop_day",
    "starts_at_home_purpose", "ends_at_home_purpose", "complete_home_based_chain",
    "n_unique_visited_zones", "n_work_school_zones", "n_errand_shop_zones",
    "n_transit_compatible_zones"
]
for c in chain_numeric_fill0:
    if c in analytic.columns:
        analytic[c] = analytic[c].fillna(0)

for c in ["visited_zones", "work_school_zones", "errand_shop_zones", "social_zones", "transit_compatible_zones"]:
    if c in analytic.columns:
        analytic[c] = analytic[c].apply(lambda x: x if isinstance(x, list) else [])

zone_lookup = zone_opp.set_index("cms_zone")["adl_opportunity_score"].to_dict()

def max_zone_score(zones):
    if not isinstance(zones, list) or len(zones) == 0:
        return 0.0
    vals = []
    for z in zones:
        try:
            vals.append(float(zone_lookup.get(int(z), 0.0)))
        except Exception:
            pass
    return max(vals) if vals else 0.0

analytic["home_zone_num"] = clean_numeric_codes(analytic["home_cms_zone"]) if "home_cms_zone" in analytic.columns else np.nan
analytic["home_zone_opp_score"] = analytic["home_zone_num"].apply(lambda z: float(zone_lookup.get(int(z), 0.0)) if pd.notna(z) else 0.0)
analytic["visited_zone_opp_max"] = analytic["visited_zones"].apply(max_zone_score)
analytic["work_school_zone_opp_max"] = analytic["work_school_zones"].apply(max_zone_score)
analytic["errand_shop_zone_opp_max"] = analytic["errand_shop_zones"].apply(max_zone_score)
analytic["transit_zone_opp_max"] = analytic["transit_compatible_zones"].apply(max_zone_score)

analytic["equity_priority"] = (
    (analytic.get("zero_vehicle_hh", 0) == 1) |
    (analytic.get("low_income_hh", 0) == 1)
).astype(int)

model_df = analytic[(analytic["valid_target_response"] == 1) & analytic[target_col].notna()].copy()
model_df[target_col] = model_df[target_col].astype(int)

print("Analytic:", analytic.shape)
print("Model:", model_df.shape)
print("Target positives:", int(model_df[target_col].sum()), "rate:", model_df[target_col].mean())
print("Model columns:", sorted(model_df.columns))

model_df.to_pickle(OUTPUT_DIR / "adl_model_table_2024.pkl")
```

## 6. Supervised revealed-behavior model

This section evaluates whether activity-chain features improve ADL prediction. This is the empirical foundation for the bandit simulator.

```python
# ============================================================
# 6. Supervised model training/evaluation
# ============================================================

core_numeric = [
    "age", "num_people", "num_adults", "num_kids", "num_workers", "num_students",
    "num_vehicles", "num_bicycles", "num_scooters", "num_trips_x", "num_trips_y",
    "travel_dow", "weekend", "shop_online_bin", "telecommute_time",
    "zero_vehicle_hh", "low_income_hh", "large_apartment_hh",
    "usual_package_alt_access", "usual_package_locker_access"
]
core_numeric = [c for c in core_numeric if c in model_df.columns]

core_categorical = [
    "home_cms_zone_cat", "home_county_cat", "residence_type_cat", "residence_rent_own_cat",
    "income_broad_cat", "gender_cat", "employment_cat", "student_cat",
    "work_in_region_cat", "work_cms_zone_cat", "school_type_cat",
    "school_in_region_cat", "school_cms_zone_cat", "telework_freq_cat",
    "commute_freq_cat", "packages_cat"
]
core_categorical = [c for c in core_categorical if c in model_df.columns]

chain_features = [
    "linked_trip_count", "total_distance_miles", "total_duration_minutes",
    "travel_span_minutes", "max_linked_trip_distance", "max_linked_trip_duration",
    "n_unique_visited_zones", "has_work_school_anchor", "has_errand_shop_anchor",
    "has_social_anchor", "has_change_mode_anchor", "has_walk_day",
    "has_bike_scooter_day", "has_auto_day", "has_taxi_tnc_day", "has_transit_day",
    "work_school_trip_count", "errand_shop_trip_count", "social_trip_count",
    "change_mode_trip_count", "multi_zone_day", "multi_stop_day",
    "complete_home_based_chain", "starts_at_home_purpose", "ends_at_home_purpose"
]
chain_features = [c for c in chain_features if c in model_df.columns]

opportunity_features = [
    "home_zone_opp_score", "visited_zone_opp_max", "work_school_zone_opp_max",
    "errand_shop_zone_opp_max", "transit_zone_opp_max",
    "n_work_school_zones", "n_errand_shop_zones", "n_transit_compatible_zones"
]
opportunity_features = [c for c in opportunity_features if c in model_df.columns]

# The supervised model is a response layer for the recommendation simulator.
# We still run ablations, but the paper's main contribution is policy evaluation rather than pure prediction.
feature_sets = {
    "M0_core_behavior": core_numeric + core_categorical,
    "M1_core_opportunity": core_numeric + core_categorical + opportunity_features,
    "M2_full_policy_context": core_numeric + core_categorical + chain_features + opportunity_features,
}
for name, feats in feature_sets.items():
    print(name, len(feats))

def make_preprocessor(features):
    cat_cols = [c for c in features if str(model_df[c].dtype) in ["category", "object", "string"] or c.endswith("_cat")]
    num_cols = [c for c in features if c not in cat_cols]
    try:
        ohe = OneHotEncoder(handle_unknown="ignore", sparse_output=True, min_frequency=10)
    except TypeError:
        ohe = OneHotEncoder(handle_unknown="ignore", sparse=True)

    pre = ColumnTransformer([
        ("num", Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler(with_mean=False))
        ]), num_cols),
        ("cat", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("onehot", ohe)
        ]), cat_cols)
    ], remainder="drop")
    return pre

def build_models(pre):
    # Default keeps the response layer fast. Heavy tree baselines can be enabled for final robustness.
    models = {
        "logit_balanced": Pipeline([
            ("prep", pre),
            ("clf", LogisticRegression(max_iter=2000, class_weight="balanced", solver="liblinear", random_state=RANDOM_STATE))
        ]),
        "extra_trees_balanced": Pipeline([
            ("prep", pre),
            ("clf", ExtraTreesClassifier(
                n_estimators=TREE_N_ESTIMATORS,
                min_samples_leaf=5,
                max_features="sqrt",
                class_weight="balanced",
                random_state=RANDOM_STATE,
                n_jobs=-1
            ))
        ]),
    }
    if RUN_HEAVY_SUPERVISED_MODELS:
        models.update({
            "random_forest_balanced": Pipeline([
                ("prep", pre),
                ("clf", RandomForestClassifier(
                    n_estimators=TREE_N_ESTIMATORS,
                    min_samples_leaf=5,
                    max_features="sqrt",
                    class_weight="balanced_subsample",
                    random_state=RANDOM_STATE,
                    n_jobs=-1
                ))
            ]),
            "gradient_boosting": Pipeline([
                ("prep", pre),
                ("clf", GradientBoostingClassifier(random_state=RANDOM_STATE))
            ]),
        })
    return models

def evaluate_prob_model(y_true, p_pred, label, sample_weight=None):
    y_true = np.asarray(y_true).astype(int)
    p_pred = np.asarray(p_pred)
    out = {
        "model": label,
        "n": len(y_true),
        "positive_rate": y_true.mean(),
        "roc_auc": safe_auc(y_true, p_pred, "roc"),
        "pr_auc": safe_auc(y_true, p_pred, "pr"),
        "brier": brier_score_loss(y_true, p_pred, sample_weight=sample_weight),
    }
    try:
        out["log_loss"] = log_loss(y_true, np.clip(p_pred, 1e-6, 1 - 1e-6), sample_weight=sample_weight)
    except Exception:
        out["log_loss"] = np.nan
    return out

def topk_metrics(y, p, ks=(0.01, 0.05, 0.10, 0.20)):
    y = np.asarray(y).astype(int)
    p = np.asarray(p)
    order = np.argsort(-p)
    rows = []
    total_pos = max(y.sum(), 1)
    base_rate = y.mean()
    for k in ks:
        n = max(1, int(np.ceil(k * len(y))))
        idx = order[:n]
        precision = y[idx].mean()
        recall = y[idx].sum() / total_pos
        lift = precision / base_rate if base_rate > 0 else np.nan
        rows.append({"top_k": k, "n": n, "precision": precision, "recall_capture": recall, "lift": lift})
    return pd.DataFrame(rows)

groups = model_df["hh_id"].astype(str)
gss = GroupShuffleSplit(n_splits=1, test_size=0.25, random_state=RANDOM_STATE)
train_idx, test_idx = next(gss.split(model_df, model_df[target_col], groups=groups))
train_df = model_df.iloc[train_idx].copy()
test_df = model_df.iloc[test_idx].copy()

print("Train:", train_df.shape, "positive rate:", train_df[target_col].mean(), "positive n:", int(train_df[target_col].sum()))
print("Test:", test_df.shape, "positive rate:", test_df[target_col].mean(), "positive n:", int(test_df[target_col].sum()))
print("Household overlap:", len(set(train_df.hh_id).intersection(set(test_df.hh_id))))

w_test = clean_numeric_codes(test_df["day_weight_citywide"]) if "day_weight_citywide" in test_df.columns else None

model_results = []
trained_models = {}

for fs_name, features in feature_sets.items():
    X_train = train_df[features].copy()
    y_train = train_df[target_col].astype(int)
    X_test = test_df[features].copy()
    y_test = test_df[target_col].astype(int)

    models = build_models(make_preprocessor(features))
    for model_name, pipe in models.items():
        label = f"{fs_name}__{model_name}"
        print("Training:", label)
        try:
            pipe.fit(X_train, y_train)
            p_test = pipe.predict_proba(X_test)[:, 1]
            res = evaluate_prob_model(y_test, p_test, label, sample_weight=w_test)
            res["feature_set"] = fs_name
            res["algorithm"] = model_name
            model_results.append(res)
            trained_models[label] = {"pipe": pipe, "features": features, "p_test": p_test}
        except Exception as e:
            print("Failed:", label, e)

model_results = pd.DataFrame(model_results).sort_values(["pr_auc", "brier"], ascending=[False, True])
display(model_results)
model_results.to_csv(OUTPUT_DIR / "supervised_model_results.csv", index=False)

best_label = model_results.iloc[0]["model"]
best_model = trained_models[best_label]["pipe"]
best_features = trained_models[best_label]["features"]
best_p_test = trained_models[best_label]["p_test"]
print("Best model:", best_label)
```

```python
# ============================================================
# 6.1 Calibration, lift, importance, zone holdout
# ============================================================

y_test = test_df[target_col].astype(int).values

try:
    calibrated = CalibratedClassifierCV(best_model, method="sigmoid", cv=3)
    calibrated.fit(train_df[best_features], train_df[target_col].astype(int))
    p_behavior = calibrated.predict_proba(test_df[best_features])[:, 1]
    behavior_model = calibrated
    print("Using sigmoid-calibrated behavior model.")
except Exception as e:
    print("Calibration failed; using uncalibrated model:", e)
    p_behavior = best_p_test
    behavior_model = best_model

cal_eval = evaluate_prob_model(y_test, p_behavior, best_label + "__behavior_model", sample_weight=w_test)
print(cal_eval)

pred_df = test_df[["hh_id", "person_id", "day_id", target_col, "equity_priority"]].copy()
pred_df["p_behavior"] = p_behavior
pred_df.to_csv(OUTPUT_DIR / "heldout_behavior_predictions.csv", index=False)

lift_df = topk_metrics(y_test, p_behavior)
display(lift_df)
lift_df.to_csv(OUTPUT_DIR / "topk_lift_behavior_model.csv", index=False)

precision, recall, _ = precision_recall_curve(y_test, p_behavior)
fpr, tpr, _ = roc_curve(y_test, p_behavior)
prob_true, prob_pred = calibration_curve(y_test, p_behavior, n_bins=10, strategy="quantile")

fig, ax = plt.subplots(figsize=(6, 5))
ax.plot(recall, precision)
ax.set_xlabel("Recall")
ax.set_ylabel("Precision")
ax.set_title("Precision-Recall Curve")
fig.tight_layout()
fig.savefig(FIG_DIR / "behavior_pr_curve.png", dpi=200)
plt.show()

fig, ax = plt.subplots(figsize=(6, 5))
ax.plot(fpr, tpr)
ax.plot([0, 1], [0, 1], linestyle="--")
ax.set_xlabel("False Positive Rate")
ax.set_ylabel("True Positive Rate")
ax.set_title("ROC Curve")
fig.tight_layout()
fig.savefig(FIG_DIR / "behavior_roc_curve.png", dpi=200)
plt.show()

fig, ax = plt.subplots(figsize=(6, 5))
ax.plot(prob_pred, prob_true, marker="o")
ax.plot([0, 1], [0, 1], linestyle="--")
ax.set_xlabel("Mean predicted probability")
ax.set_ylabel("Observed frequency")
ax.set_title("Calibration Curve")
fig.tight_layout()
fig.savefig(FIG_DIR / "behavior_calibration_curve.png", dpi=200)
plt.show()

if RUN_PERMUTATION_IMPORTANCE:
    try:
        result = permutation_importance(
            behavior_model, test_df[best_features], y_test,
            n_repeats=5, random_state=RANDOM_STATE,
            scoring="average_precision"
        )
        imp = pd.DataFrame({
            "feature": best_features,
            "importance_mean": result.importances_mean,
            "importance_std": result.importances_std
        }).sort_values("importance_mean", ascending=False)
        display(imp.head(25))
        imp.to_csv(OUTPUT_DIR / "permutation_importance_average_precision.csv", index=False)
    except Exception as e:
        print("Permutation importance failed:", e)

if RUN_ZONE_HOLDOUT and "home_zone_num" in model_df.columns:
    rows = []
    features = feature_sets["M2_full_policy_context"]
    zones = model_df["home_zone_num"].fillna(-999).astype(int)
    logo = LeaveOneGroupOut()

    for tr, te in logo.split(model_df, model_df[target_col], groups=zones):
        held_zone = zones.iloc[te].iloc[0]
        if len(np.unique(model_df.iloc[te][target_col])) < 2 or len(np.unique(model_df.iloc[tr][target_col])) < 2:
            continue
        pipe = Pipeline([
            ("prep", make_preprocessor(features)),
            ("clf", LogisticRegression(max_iter=2000, class_weight="balanced", solver="liblinear", random_state=RANDOM_STATE))
        ])
        try:
            pipe.fit(model_df.iloc[tr][features], model_df.iloc[tr][target_col].astype(int))
            p = pipe.predict_proba(model_df.iloc[te][features])[:, 1]
            res = evaluate_prob_model(model_df.iloc[te][target_col].astype(int), p, f"holdout_zone_{held_zone}")
            res["heldout_zone"] = held_zone
            res["n_pos"] = int(model_df.iloc[te][target_col].sum())
            rows.append(res)
        except Exception as e:
            print("Zone holdout failed:", held_zone, e)

    zone_holdout = pd.DataFrame(rows)
    display(zone_holdout)
    zone_holdout.to_csv(OUTPUT_DIR / "zone_holdout_results.csv", index=False)
```

## 7. Action-specific OOH/ADL recommendation environment

The simulator expands each person-day into six transparent candidate actions:

1. `HOME_DELIVERY`
2. `HOME_ZONE_ADL`
3. `WORK_SCHOOL_ADL`
4. `ERRAND_SHOP_ADL`
5. `TRANSIT_ADL`
6. `MIN_BURDEN_VISITED_ADL`

Eligibility comes from observed activity-chain compatibility and a zone opportunity score. Reward combines assumed acceptance utility, opportunity, service value, burden, and equity-related terms. These coefficients encode a research scenario; they are not estimates of causal action effects.

In the included run, the opportunity measure is a survey-derived proxy. The reward design strongly favors the minimum-burden visited-zone action, which ties the simulator oracle. Interpret the policy results as diagnostics of this constructed environment, not evidence that one delivery policy is optimal in practice.

```python
# ============================================================
# 7. Action environment: OOH/ADL recommendation simulator
# ============================================================

ACTIONS = [
    "HOME_DELIVERY",
    "HOME_ZONE_ADL",
    "WORK_SCHOOL_ADL",
    "ERRAND_SHOP_ADL",
    "TRANSIT_ADL",
    "MIN_BURDEN_VISITED_ADL",
]
ACTION_TO_ID = {a: i for i, a in enumerate(ACTIONS)}

# Action-level burden and service assumptions. These are deliberately transparent and varied later.
ACTION_BASE_BURDEN = {
    "HOME_DELIVERY": 0.00,
    "HOME_ZONE_ADL": 0.32,
    "WORK_SCHOOL_ADL": 0.14,
    "ERRAND_SHOP_ADL": 0.10,
    "TRANSIT_ADL": 0.18,
    "MIN_BURDEN_VISITED_ADL": 0.08,
}
ACTION_SERVICE_VALUE = {
    "HOME_DELIVERY": 0.00,
    "HOME_ZONE_ADL": 0.12,
    "WORK_SCHOOL_ADL": 0.22,
    "ERRAND_SHOP_ADL": 0.24,
    "TRANSIT_ADL": 0.20,
    "MIN_BURDEN_VISITED_ADL": 0.26,
}

def best_zone(zones, zone_score_lookup):
    zones = zones if isinstance(zones, list) else []
    valid = []
    for z in zones:
        if pd.notna(z):
            try:
                zi = int(z)
                valid.append((zi, float(zone_score_lookup.get(zi, 0.0))))
            except Exception:
                pass
    if not valid:
        return np.nan, 0.0
    return max(valid, key=lambda x: x[1])

def make_action_features_for_row(row, zone_score_lookup):
    home_zone = row.get("home_zone_num", np.nan)
    visited = row.get("visited_zones", [])
    work_school = row.get("work_school_zones", [])
    errand = row.get("errand_shop_zones", [])
    transit = row.get("transit_compatible_zones", [])
    out = []

    out.append({
        "action": "HOME_DELIVERY",
        "action_id": ACTION_TO_ID["HOME_DELIVERY"],
        "candidate_zone": home_zone,
        "opp_score": 0.0,
        "available": 1,
        "eligible": 1,
        "chain_compatible": 1,
        "requires_extra_trip": 0,
        "burden": 0.0,
        "is_adl_action": 0,
        "service_value": ACTION_SERVICE_VALUE["HOME_DELIVERY"],
    })

    def add(action, candidate_zones, extra_trip, chain_compatible):
        z, opp = best_zone(candidate_zones, zone_score_lookup)
        available = int(opp > 0)
        eligible = int(available and chain_compatible)
        burden = ACTION_BASE_BURDEN[action]
        burden += 0.25 * (1 - available)
        burden += 0.18 * int(extra_trip)
        # Slightly reduce burden when opportunity is high; this represents richer local ADL supply.
        burden -= 0.10 * float(opp)
        burden = float(np.clip(burden, 0.01, 2.0))
        out.append({
            "action": action,
            "action_id": ACTION_TO_ID[action],
            "candidate_zone": z,
            "opp_score": float(opp),
            "available": available,
            "eligible": eligible,
            "chain_compatible": int(chain_compatible),
            "requires_extra_trip": int(extra_trip),
            "burden": burden,
            "is_adl_action": 1,
            "service_value": ACTION_SERVICE_VALUE[action],
        })

    home_zones = [int(home_zone)] if pd.notna(home_zone) else []
    add("HOME_ZONE_ADL", home_zones, extra_trip=1, chain_compatible=1)
    add("WORK_SCHOOL_ADL", work_school, extra_trip=0, chain_compatible=int(len(work_school) > 0))
    add("ERRAND_SHOP_ADL", errand, extra_trip=0, chain_compatible=int(len(errand) > 0))
    add("TRANSIT_ADL", transit, extra_trip=0, chain_compatible=int(len(transit) > 0))
    add("MIN_BURDEN_VISITED_ADL", visited, extra_trip=0, chain_compatible=int(len(visited) > 0))
    return out

def build_action_table(sim_df, p_base, zone_score_lookup):
    rows = []
    for obs_idx, (_, row) in enumerate(sim_df.iterrows()):
        for a in make_action_features_for_row(row, zone_score_lookup):
            a["obs_id"] = obs_idx
            a["hh_id"] = row["hh_id"]
            a["person_id"] = row["person_id"]
            a["day_id"] = row["day_id"]
            a["p_base"] = float(p_base[obs_idx])
            a["equity_priority"] = int(row.get("equity_priority", 0))
            rows.append(a)
    return pd.DataFrame(rows)

sim_df = test_df.copy().reset_index(drop=True)
p_sim_base = p_behavior.copy()
if MAX_BANDIT_EVENTS is not None:
    sample_idx = sim_df.sample(n=min(MAX_BANDIT_EVENTS, len(sim_df)), random_state=RANDOM_STATE).index
    sim_df = sim_df.loc[sample_idx].reset_index(drop=True)
    p_sim_base = p_behavior[sample_idx]

action_table = build_action_table(sim_df, p_sim_base, zone_lookup)
display(action_table.groupby("action")[["available", "eligible", "burden", "opp_score"]].mean())
print("Action table:", action_table.shape)
action_table.to_csv(OUTPUT_DIR / "action_table_heldout.csv", index=False)

# Richer environment: opportunity and service value can overcome modest burden differences.
ENV_PARAMS_DEFAULT = {
    "beta_opp": 0.90,
    "beta_burden": 0.85,
    "beta_chain": 0.55,
    "beta_available": 0.45,
    "beta_extra_trip": 0.45,
    "success_value": 1.00,
    "service_value_weight": 0.60,
    "opportunity_reward": 0.35,
    "consolidation_reward": 0.20,
    "behavior_targeting_reward": 0.15,
    "burden_penalty": 0.22,
    "equity_bonus": 0.18,
    "equity_burden_penalty": 0.10,
    "recommendation_cost": 0.015,
    "ineligible_penalty": 0.20,
}

def calibrate_action_intercept(action_table, params, target_mean, baseline_action="MIN_BURDEN_VISITED_ADL"):
    sub = action_table[action_table["action"] == baseline_action].copy()
    base_logit = logit(sub["p_base"].values)
    z_no_int = (
        base_logit
        + params["beta_opp"] * sub["opp_score"].values
        - params["beta_burden"] * sub["burden"].values
        + params["beta_chain"] * sub["chain_compatible"].values
        + params["beta_available"] * sub["available"].values
        - params["beta_extra_trip"] * sub["requires_extra_trip"].values
    )
    lo, hi = -10, 10
    for _ in range(80):
        mid = (lo + hi) / 2
        if sigmoid(z_no_int + mid).mean() < target_mean:
            lo = mid
        else:
            hi = mid
    return (lo + hi) / 2

def add_environment_columns(action_table, params=None, target_mean=None):
    params = dict(ENV_PARAMS_DEFAULT if params is None else params)
    at = action_table.copy()
    if target_mean is None:
        target_mean = float(sim_df[target_col].mean())

    intercept = calibrate_action_intercept(at, params, target_mean)
    at["env_intercept"] = intercept

    util = (
        logit(at["p_base"].values)
        + intercept
        + params["beta_opp"] * at["opp_score"].values
        - params["beta_burden"] * at["burden"].values
        + params["beta_chain"] * at["chain_compatible"].values
        + params["beta_available"] * at["available"].values
        - params["beta_extra_trip"] * at["requires_extra_trip"].values
    )
    p_accept = sigmoid(util)
    p_accept = p_accept * at["is_adl_action"].values * at["eligible"].values
    at["p_accept"] = np.clip(p_accept, 0, 1)

    # Reward components. Expected reward = P(accept)*accept_reward + deterministic_cost.
    at["accept_reward"] = (
        params["success_value"]
        + params["service_value_weight"] * at["service_value"]
        + params["opportunity_reward"] * at["opp_score"]
        + params["consolidation_reward"] * at["opp_score"] * at["chain_compatible"]
        + params["behavior_targeting_reward"] * at["p_base"]
        + params["equity_bonus"] * at["equity_priority"]
    )
    at["deterministic_cost"] = (
        - params["burden_penalty"] * at["burden"] * at["is_adl_action"]
        - params["equity_burden_penalty"] * at["burden"] * at["equity_priority"] * at["is_adl_action"]
        - params["recommendation_cost"] * at["is_adl_action"]
        - params["ineligible_penalty"] * ((at["is_adl_action"] == 1) & (at["eligible"] == 0)).astype(int)
    )
    at["expected_reward"] = at["p_accept"] * at["accept_reward"] + at["deterministic_cost"]
    at.loc[at["action"] == "HOME_DELIVERY", ["p_accept", "accept_reward", "deterministic_cost", "expected_reward"]] = [0.0, 0.0, 0.0, 0.0]
    return at, intercept

action_env, env_intercept = add_environment_columns(action_table, ENV_PARAMS_DEFAULT)
print("Calibrated action intercept:", env_intercept)
display(action_env.groupby("action")[["p_accept", "expected_reward", "accept_reward", "deterministic_cost", "burden", "opp_score", "eligible"]].mean())
action_env.to_csv(OUTPUT_DIR / "action_environment_heldout.csv", index=False)

fig, ax = plt.subplots(figsize=(8, 4))
action_env.groupby("action")["p_accept"].mean().reindex(ACTIONS).plot(kind="bar", ax=ax)
ax.set_ylabel("Mean simulated OOH acceptance probability")
ax.set_title("Simulated acceptance by recommendation action")
ax.tick_params(axis="x", rotation=35)
fig.tight_layout()
fig.savefig(FIG_DIR / "simulated_acceptance_by_action.png", dpi=200)
plt.show()

fig, ax = plt.subplots(figsize=(8, 4))
action_env.groupby("action")["expected_reward"].mean().reindex(ACTIONS).plot(kind="bar", ax=ax)
ax.set_ylabel("Mean expected reward")
ax.set_title("Expected reward by recommendation action")
ax.tick_params(axis="x", rotation=35)
fig.tight_layout()
fig.savefig(FIG_DIR / "expected_reward_by_action.png", dpi=200)
plt.show()
```

## 8. Static policy benchmarks

Strong baselines are essential. The contextual bandit should be compared with practical static rules, not only random choice. This section separates:

- **implementable static policies**: nearest/home-zone, activity-priority, high-propensity, equity-priority, highest opportunity;
- **learning policies**: contextual bandits in later cells;
- **upper bound**: `ORACLE_EXPECTED_REWARD`, which knows the simulator's expected reward and is not implementable in practice.


```python
# ============================================================
# 8. Static policy evaluation with stronger baselines
# ============================================================

def env_to_matrices(action_env):
    mats = {}
    matrix_cols = [
        "p_accept", "expected_reward", "accept_reward", "deterministic_cost",
        "burden", "eligible", "opp_score", "equity_priority", "p_base",
        "chain_compatible", "requires_extra_trip", "is_adl_action"
    ]
    for col in matrix_cols:
        mats[col] = action_env.pivot(index="obs_id", columns="action_id", values=col).sort_index(axis=1).values
    return mats

env_mats = env_to_matrices(action_env)

def fallback_home_if_bad(choices, mats):
    choices = np.asarray(choices).astype(int).copy()
    idx = np.arange(len(choices))
    bad = (mats["eligible"][idx, choices] < 1) | (mats["expected_reward"][idx, choices] < 0)
    choices[bad] = ACTION_TO_ID["HOME_DELIVERY"]
    return choices

def static_policy_choices(policy_name, mats, propensity_threshold=None):
    n, a = mats["expected_reward"].shape
    home = ACTION_TO_ID["HOME_DELIVERY"]
    if propensity_threshold is None:
        # Top quartile of behavior propensity in held-out set.
        propensity_threshold = np.quantile(mats["p_base"][:, home], 0.75)

    if policy_name == "HOME_ONLY":
        return np.zeros(n, dtype=int)
    if policy_name in ACTION_TO_ID:
        return fallback_home_if_bad(np.full(n, ACTION_TO_ID[policy_name], dtype=int), mats)
    if policy_name == "HIGHEST_OPPORTUNITY":
        opp = mats["opp_score"].copy()
        opp[:, home] = -np.inf
        choices = np.argmax(opp, axis=1)
        return fallback_home_if_bad(choices, mats)
    if policy_name == "HIGHEST_PREDICTED_ACCEPTANCE":
        p = mats["p_accept"].copy()
        p[:, home] = -np.inf
        choices = np.argmax(p, axis=1)
        return fallback_home_if_bad(choices, mats)
    if policy_name == "LOWEST_BURDEN_ELIGIBLE":
        burden = mats["burden"].copy()
        eligible = mats["eligible"].copy()
        burden[:, home] = np.inf
        burden = np.where(eligible == 1, burden, np.inf)
        choices = np.argmin(burden, axis=1)
        choices[np.isinf(burden).all(axis=1)] = home
        return fallback_home_if_bad(choices, mats)
    if policy_name == "HIGH_PROPENSITY_MIN_BURDEN":
        choices = static_policy_choices("LOWEST_BURDEN_ELIGIBLE", mats, propensity_threshold)
        low_prop = mats["p_base"][:, home] < propensity_threshold
        choices[low_prop] = home
        return fallback_home_if_bad(choices, mats)
    if policy_name == "EQUITY_PRIORITY_MIN_BURDEN":
        choices = static_policy_choices("LOWEST_BURDEN_ELIGIBLE", mats, propensity_threshold)
        non_equity = mats["equity_priority"][:, home] < 1
        choices[non_equity] = home
        return fallback_home_if_bad(choices, mats)
    if policy_name == "ACTIVITY_PRIORITY_RULE":
        # Practical rule: errand if possible, then work/school, transit, visited-zone, home-zone, else home.
        priority = ["ERRAND_SHOP_ADL", "WORK_SCHOOL_ADL", "TRANSIT_ADL", "MIN_BURDEN_VISITED_ADL", "HOME_ZONE_ADL"]
        choices = np.full(n, home, dtype=int)
        for action in reversed(priority):
            aid = ACTION_TO_ID[action]
            ok = (mats["eligible"][:, aid] == 1) & (mats["expected_reward"][:, aid] >= 0)
            choices[ok] = aid
        return fallback_home_if_bad(choices, mats)
    if policy_name == "SCORE_HEURISTIC":
        # A transparent implementable scoring heuristic that does not use full expected reward.
        score = (
            0.50 * mats["p_accept"]
            + 0.25 * mats["opp_score"]
            + 0.15 * mats["chain_compatible"]
            + 0.10 * mats["equity_priority"]
            - 0.25 * mats["burden"]
        )
        score[:, home] = 0.0
        score = np.where(mats["eligible"] == 1, score, -np.inf)
        choices = np.argmax(score, axis=1)
        # If no ADL score beats home baseline, choose home.
        choices[np.nanmax(score, axis=1) <= 0] = home
        return fallback_home_if_bad(choices, mats)
    if policy_name == "ORACLE_EXPECTED_REWARD":
        choices = np.argmax(mats["expected_reward"], axis=1)
        return choices
    raise ValueError(policy_name)

def evaluate_choices(choices, mats, equity, label):
    idx = np.arange(len(choices))
    p = mats["p_accept"][idx, choices]
    r = mats["expected_reward"][idx, choices]
    burden = mats["burden"][idx, choices]
    opp = mats["opp_score"][idx, choices]
    is_adl = (choices != ACTION_TO_ID["HOME_DELIVERY"]).astype(int)
    eq = np.asarray(equity).astype(int)
    eq_mask = eq == 1
    neq_mask = eq == 0
    return {
        "policy": label,
        "n": len(choices),
        "mean_expected_reward": float(np.mean(r)),
        "mean_p_accept": float(np.mean(p)),
        "expected_ooh_acceptances": float(np.sum(p)),
        "recommend_ooh_rate": float(np.mean(is_adl)),
        "mean_opportunity_if_recommended": float(np.mean(opp[is_adl == 1])) if is_adl.sum() > 0 else 0.0,
        "mean_burden_if_recommended": float(np.mean(burden[is_adl == 1])) if is_adl.sum() > 0 else 0.0,
        "expected_accepted_burden": float(np.sum(p * burden) / max(np.sum(p), 1e-9)),
        "equity_group_share": float(eq.mean()),
        "equity_expected_accept_rate": float(np.mean(p[eq_mask])) if eq_mask.any() else np.nan,
        "non_equity_expected_accept_rate": float(np.mean(p[neq_mask])) if neq_mask.any() else np.nan,
        "equity_service_gap": float(np.mean(p[eq_mask]) - np.mean(p[neq_mask])) if eq_mask.any() and neq_mask.any() else np.nan,
        "equity_expected_burden": float(np.sum(p[eq_mask] * burden[eq_mask]) / max(np.sum(p[eq_mask]), 1e-9)) if eq_mask.any() else np.nan,
        "non_equity_expected_burden": float(np.sum(p[neq_mask] * burden[neq_mask]) / max(np.sum(p[neq_mask]), 1e-9)) if neq_mask.any() else np.nan,
    }

static_policy_names = [
    "HOME_ONLY", "HOME_ZONE_ADL", "WORK_SCHOOL_ADL", "ERRAND_SHOP_ADL",
    "TRANSIT_ADL", "MIN_BURDEN_VISITED_ADL", "HIGHEST_OPPORTUNITY",
    "HIGHEST_PREDICTED_ACCEPTANCE", "LOWEST_BURDEN_ELIGIBLE",
    "HIGH_PROPENSITY_MIN_BURDEN", "EQUITY_PRIORITY_MIN_BURDEN",
    "ACTIVITY_PRIORITY_RULE", "SCORE_HEURISTIC", "ORACLE_EXPECTED_REWARD"
]

equity = env_mats["equity_priority"][:, 0]
static_rows = []
static_choices = {}
for pol in static_policy_names:
    choices = static_policy_choices(pol, env_mats)
    static_choices[pol] = choices
    static_rows.append(evaluate_choices(choices, env_mats, equity, pol))

static_results = pd.DataFrame(static_rows).sort_values("mean_expected_reward", ascending=False)
display(static_results)
static_results.to_csv(OUTPUT_DIR / "static_policy_results.csv", index=False)

def bootstrap_static_policies(static_choices, mats, equity, n_boot=300, seed=RANDOM_STATE):
    rng_local = np.random.default_rng(seed)
    n = len(equity)
    rows = []
    for b in range(n_boot):
        sample_idx = rng_local.integers(0, n, size=n)
        sub_mats = {k: v[sample_idx] for k, v in mats.items()}
        sub_eq = equity[sample_idx]
        for pol, choices in static_choices.items():
            sub_choices = choices[sample_idx]
            res = evaluate_choices(sub_choices, sub_mats, sub_eq, pol)
            res["bootstrap"] = b
            rows.append(res)
    boot = pd.DataFrame(rows)
    summary = []
    metrics = ["mean_expected_reward", "mean_p_accept", "recommend_ooh_rate", "expected_accepted_burden", "equity_service_gap"]
    for pol, g in boot.groupby("policy"):
        for m in metrics:
            summary.append({
                "policy": pol,
                "metric": m,
                "mean": g[m].mean(),
                "ci_low": g[m].quantile(0.025),
                "ci_high": g[m].quantile(0.975)
            })
    return pd.DataFrame(summary)

static_ci = bootstrap_static_policies(static_choices, env_mats, equity, n_boot=300)
display(static_ci.head())
static_ci.to_csv(OUTPUT_DIR / "static_policy_bootstrap_ci.csv", index=False)

fig, ax = plt.subplots(figsize=(10, 5))
plot_df = static_results[static_results["policy"] != "ORACLE_EXPECTED_REWARD"].sort_values("mean_expected_reward", ascending=False)
ax.bar(plot_df["policy"], plot_df["mean_expected_reward"])
ax.set_ylabel("Mean expected reward")
ax.set_title("Implementable static recommendation policies")
ax.tick_params(axis="x", rotation=45, labelsize=8)
fig.tight_layout()
fig.savefig(FIG_DIR / "static_policy_expected_reward.png", dpi=200)
plt.show()
```

## 9. Contextual bandit algorithms

Implemented policies:

- Epsilon-greedy linear bandit
- LinUCB
- Linear Thompson Sampling
- Constrained LinUCB with an equity-priority burden filter

```python
# ============================================================
# 9. Bandit features and algorithms
# ============================================================

def build_bandit_feature_matrices(sim_df, action_env):
    context_cols = [
        "p_behavior", "equity_priority", "zero_vehicle_hh", "low_income_hh",
        "large_apartment_hh", "age", "num_people", "num_workers", "num_students",
        "num_vehicles", "travel_dow", "weekend", "shop_online_bin",
        "linked_trip_count", "total_distance_miles", "total_duration_minutes",
        "n_unique_visited_zones", "has_work_school_anchor", "has_errand_shop_anchor",
        "has_transit_day", "has_walk_day", "has_auto_day", "multi_zone_day", "multi_stop_day",
        "home_zone_opp_score", "visited_zone_opp_max", "work_school_zone_opp_max",
        "errand_shop_zone_opp_max", "transit_zone_opp_max"
    ]
    sc = sim_df.copy()
    sc["p_behavior"] = p_sim_base
    context_cols = [c for c in context_cols if c in sc.columns]

    C = sc[context_cols].copy()
    for c in C.columns:
        C[c] = clean_numeric_codes(C[c]).astype(float)
    C = C.replace([np.inf, -np.inf], np.nan).fillna(C.median(numeric_only=True)).fillna(0.0)

    C_values = C.values.astype(float)
    C_mean = C_values.mean(axis=0)
    C_std = C_values.std(axis=0)
    C_std[C_std == 0] = 1
    C_scaled = (C_values - C_mean) / C_std

    n = len(sim_df)
    n_actions = len(ACTIONS)

    action_cols = ["opp_score", "burden", "available", "eligible", "chain_compatible", "requires_extra_trip", "is_adl_action"]
    Afeat = np.stack([
        action_env.pivot(index="obs_id", columns="action_id", values=c).sort_index(axis=1).values.astype(float)
        for c in action_cols
    ], axis=2)

    action_onehot = np.eye(n_actions)[np.arange(n_actions)]
    action_onehot = np.broadcast_to(action_onehot, (n, n_actions, n_actions))
    C3 = np.broadcast_to(C_scaled[:, None, :], (n, n_actions, C_scaled.shape[1]))

    X = np.concatenate([C3, Afeat, action_onehot], axis=2)
    names = [f"ctx_{c}" for c in context_cols] + [f"act_{c}" for c in action_cols] + [f"action_{a}" for a in ACTIONS]
    return X.astype(float), names

X_bandit, bandit_feature_names = build_bandit_feature_matrices(sim_df, action_env)
P_accept = env_mats["p_accept"]
Expected_R = env_mats["expected_reward"]
AcceptReward = env_mats["accept_reward"]
DeterministicCost = env_mats["deterministic_cost"]
Burden = env_mats["burden"]
Eligible = env_mats["eligible"]
Equity = env_mats["equity_priority"][:, 0].astype(int)

print("Bandit X:", X_bandit.shape)

class BaseBanditPolicy:
    def __init__(self, n_actions, d, seed=RANDOM_STATE, name="base"):
        self.n_actions = n_actions
        self.d = d
        self.rng = np.random.default_rng(seed)
        self.name = name

    def select(self, X_t, t, feasible_mask=None, burden_t=None, equity_t=0):
        raise NotImplementedError

    def update(self, action, x, reward):
        pass

class EpsilonGreedyLinear(BaseBanditPolicy):
    def __init__(self, n_actions, d, epsilon=0.10, ridge=1.0, seed=RANDOM_STATE):
        super().__init__(n_actions, d, seed, name=f"EpsilonGreedyLinear_eps{epsilon}")
        self.epsilon = epsilon
        self.A = np.array([ridge * np.eye(d) for _ in range(n_actions)])
        self.b = np.zeros((n_actions, d))

    def theta(self, a):
        return np.linalg.solve(self.A[a], self.b[a])

    def select(self, X_t, t, feasible_mask=None, burden_t=None, equity_t=0):
        feasible = np.ones(self.n_actions, dtype=bool) if feasible_mask is None else feasible_mask.copy()
        candidates = np.where(feasible)[0]
        if len(candidates) == 0:
            return 0
        if self.rng.random() < self.epsilon:
            return int(self.rng.choice(candidates))
        scores = np.array([X_t[a] @ self.theta(a) for a in range(self.n_actions)])
        scores[~feasible] = -np.inf
        return int(np.argmax(scores))

    def update(self, action, x, reward):
        self.A[action] += np.outer(x, x)
        self.b[action] += reward * x

class LinUCB(BaseBanditPolicy):
    def __init__(self, n_actions, d, alpha=0.75, ridge=1.0, seed=RANDOM_STATE, name=None):
        super().__init__(n_actions, d, seed, name=name or f"LinUCB_alpha{alpha}")
        self.alpha = alpha
        self.A = np.array([ridge * np.eye(d) for _ in range(n_actions)])
        self.b = np.zeros((n_actions, d))

    def select(self, X_t, t, feasible_mask=None, burden_t=None, equity_t=0):
        feasible = np.ones(self.n_actions, dtype=bool) if feasible_mask is None else feasible_mask.copy()
        scores = np.full(self.n_actions, -np.inf)
        for a in range(self.n_actions):
            if not feasible[a]:
                continue
            A_inv = np.linalg.inv(self.A[a])
            theta = A_inv @ self.b[a]
            x = X_t[a]
            scores[a] = x @ theta + self.alpha * np.sqrt(x @ A_inv @ x)
        return int(np.argmax(scores))

    def update(self, action, x, reward):
        self.A[action] += np.outer(x, x)
        self.b[action] += reward * x

class LinearThompsonSampling(BaseBanditPolicy):
    def __init__(self, n_actions, d, v=0.30, ridge=1.0, seed=RANDOM_STATE):
        super().__init__(n_actions, d, seed, name=f"LinearTS_v{v}")
        self.v = v
        self.A = np.array([ridge * np.eye(d) for _ in range(n_actions)])
        self.b = np.zeros((n_actions, d))

    def select(self, X_t, t, feasible_mask=None, burden_t=None, equity_t=0):
        feasible = np.ones(self.n_actions, dtype=bool) if feasible_mask is None else feasible_mask.copy()
        scores = np.full(self.n_actions, -np.inf)
        for a in range(self.n_actions):
            if not feasible[a]:
                continue
            A_inv = np.linalg.inv(self.A[a])
            theta_hat = A_inv @ self.b[a]
            # Diagonal Gaussian approximation avoids expensive SVD per decision.
            theta_sample = theta_hat + self.v * self.rng.normal(size=self.d) * np.sqrt(np.maximum(np.diag(A_inv), 1e-12))
            scores[a] = X_t[a] @ theta_sample
        return int(np.argmax(scores))

    def update(self, action, x, reward):
        self.A[action] += np.outer(x, x)
        self.b[action] += reward * x

class ConstrainedLinUCB(LinUCB):
    def __init__(self, n_actions, d, alpha=0.75, ridge=1.0, equity_max_burden=0.45, burden_penalty=0.20, seed=RANDOM_STATE):
        super().__init__(n_actions, d, alpha=alpha, ridge=ridge, seed=seed, name=f"ConstrainedLinUCB_b{equity_max_burden}")
        self.equity_max_burden = equity_max_burden
        self.burden_penalty = burden_penalty

    def select(self, X_t, t, feasible_mask=None, burden_t=None, equity_t=0):
        feasible = np.ones(self.n_actions, dtype=bool) if feasible_mask is None else feasible_mask.copy()
        if equity_t == 1 and burden_t is not None:
            high_burden_adl = (burden_t > self.equity_max_burden) & (np.arange(self.n_actions) != ACTION_TO_ID["HOME_DELIVERY"])
            feasible[high_burden_adl] = False
        if feasible.sum() == 0:
            feasible[ACTION_TO_ID["HOME_DELIVERY"]] = True

        scores = np.full(self.n_actions, -np.inf)
        for a in range(self.n_actions):
            if not feasible[a]:
                continue
            A_inv = np.linalg.inv(self.A[a])
            theta = A_inv @ self.b[a]
            x = X_t[a]
            penalty = self.burden_penalty * burden_t[a] if burden_t is not None else 0
            scores[a] = x @ theta + self.alpha * np.sqrt(x @ A_inv @ x) - penalty
        return int(np.argmax(scores))
```

```python
# ============================================================
# 9.1 Run bandit simulations
# ============================================================

def run_bandit_once(policy, X, P_accept, Expected_R, AcceptReward, DeterministicCost, Burden, Eligible, Equity, seed=RANDOM_STATE):
    rng_local = np.random.default_rng(seed)
    n, n_actions, d = X.shape
    order = rng_local.permutation(n)
    records = []
    cumulative_reward = 0.0
    cumulative_expected_reward = 0.0
    cumulative_oracle_expected = 0.0

    for t, i in enumerate(order):
        X_t = X[i]
        p_t = P_accept[i]
        exp_r_t = Expected_R[i]
        burden_t = Burden[i]
        feasible = Eligible[i].astype(bool)
        feasible[ACTION_TO_ID["HOME_DELIVERY"]] = True

        a = policy.select(X_t, t, feasible_mask=feasible, burden_t=burden_t, equity_t=Equity[i])

        accepted = rng_local.binomial(1, p_t[a])
        reward = accepted * AcceptReward[i, a] + DeterministicCost[i, a]
        policy.update(a, X_t[a], reward)

        expected_chosen = exp_r_t[a]
        oracle_reward = np.max(exp_r_t)
        cumulative_reward += reward
        cumulative_expected_reward += expected_chosen
        cumulative_oracle_expected += oracle_reward

        records.append({
            "t": t,
            "obs_id": int(i),
            "action_id": int(a),
            "action": ACTIONS[a],
            "accepted": int(accepted),
            "reward": float(reward),
            "expected_reward_chosen": float(expected_chosen),
            "oracle_expected_reward": float(oracle_reward),
            "instant_expected_regret": float(oracle_reward - expected_chosen),
            "cumulative_reward": float(cumulative_reward),
            "cumulative_expected_reward_chosen": float(cumulative_expected_reward),
            "cumulative_oracle_expected_reward": float(cumulative_oracle_expected),
            "cumulative_expected_regret": float(cumulative_oracle_expected - cumulative_expected_reward),
            "burden": float(burden_t[a]),
            "p_accept": float(p_t[a]),
            "equity_priority": int(Equity[i]),
        })
    return pd.DataFrame(records)

def summarize_bandit_run(run_df, policy_name, run_id):
    rec_ooh = run_df["action"] != "HOME_DELIVERY"
    eq = run_df["equity_priority"] == 1
    neq = ~eq
    return {
        "policy": policy_name,
        "run": run_id,
        "n": len(run_df),
        "mean_reward": run_df["reward"].mean(),
        "total_reward": run_df["reward"].sum(),
        "mean_expected_reward_chosen": run_df["expected_reward_chosen"].mean(),
        "mean_p_accept": run_df["p_accept"].mean(),
        "accepted_rate": run_df["accepted"].mean(),
        "recommend_ooh_rate": rec_ooh.mean(),
        "mean_burden_if_recommended": run_df.loc[rec_ooh, "burden"].mean() if rec_ooh.any() else 0.0,
        "mean_burden_if_accepted": np.average(run_df["burden"], weights=run_df["accepted"]) if run_df["accepted"].sum() > 0 else 0.0,
        "final_cumulative_expected_regret": run_df["instant_expected_regret"].sum(),
        "equity_accept_rate": run_df.loc[eq, "accepted"].mean() if eq.any() else np.nan,
        "non_equity_accept_rate": run_df.loc[neq, "accepted"].mean() if neq.any() else np.nan,
        "equity_service_gap": run_df.loc[eq, "accepted"].mean() - run_df.loc[neq, "accepted"].mean() if eq.any() and neq.any() else np.nan,
    }

def run_bandit_experiment(X, P_accept, Expected_R, AcceptReward, DeterministicCost, Burden, Eligible, Equity, n_runs=20):
    n_actions = X.shape[1]
    d = X.shape[2]
    all_runs = []
    summaries = []

    for r in range(n_runs):
        policies = [
            EpsilonGreedyLinear(n_actions, d, epsilon=0.10, seed=RANDOM_STATE + r),
            LinUCB(n_actions, d, alpha=0.75, seed=RANDOM_STATE + r),
            LinearThompsonSampling(n_actions, d, v=0.30, seed=RANDOM_STATE + r),
            ConstrainedLinUCB(n_actions, d, alpha=0.75, equity_max_burden=0.45, burden_penalty=0.25, seed=RANDOM_STATE + r),
        ]
        for pol in policies:
            print(f"Run {r+1}/{n_runs}: {pol.name}")
            run_df = run_bandit_once(
                pol, X, P_accept, Expected_R, AcceptReward, DeterministicCost,
                Burden, Eligible, Equity, seed=RANDOM_STATE + 1000*r
            )
            run_df["policy"] = pol.name
            run_df["run"] = r
            all_runs.append(run_df)
            summaries.append(summarize_bandit_run(run_df, pol.name, r))
    return pd.concat(all_runs, ignore_index=True), pd.DataFrame(summaries)

bandit_runs, bandit_summary = run_bandit_experiment(
    X_bandit, P_accept, Expected_R, AcceptReward, DeterministicCost,
    Burden, Eligible, Equity, n_runs=N_BANDIT_RUNS
)

bandit_runs.to_csv(OUTPUT_DIR / "bandit_runs.csv", index=False)
bandit_summary.to_csv(OUTPUT_DIR / "bandit_summary.csv", index=False)

bandit_policy_summary = bandit_summary.groupby("policy").agg(
    mean_reward=("mean_reward", "mean"),
    reward_sd=("mean_reward", "std"),
    mean_expected_reward_chosen=("mean_expected_reward_chosen", "mean"),
    expected_reward_sd=("mean_expected_reward_chosen", "std"),
    accepted_rate=("accepted_rate", "mean"),
    accepted_rate_sd=("accepted_rate", "std"),
    regret=("final_cumulative_expected_regret", "mean"),
    regret_sd=("final_cumulative_expected_regret", "std"),
    burden_if_accepted=("mean_burden_if_accepted", "mean"),
    equity_gap=("equity_service_gap", "mean")
).reset_index().sort_values("mean_expected_reward_chosen", ascending=False)

display(bandit_policy_summary)
bandit_policy_summary.to_csv(OUTPUT_DIR / "bandit_policy_summary.csv", index=False)

fig, ax = plt.subplots(figsize=(9, 5))
ax.bar(bandit_policy_summary["policy"], bandit_policy_summary["mean_expected_reward_chosen"],
       yerr=bandit_policy_summary["expected_reward_sd"].fillna(0))
ax.set_ylabel("Mean expected reward chosen")
ax.set_title("Contextual bandit policy performance")
ax.tick_params(axis="x", rotation=45)
fig.tight_layout()
fig.savefig(FIG_DIR / "bandit_mean_expected_reward.png", dpi=200)
plt.show()

curve = bandit_runs.groupby(["policy", "run", "t"])["cumulative_expected_reward_chosen"].last().reset_index()
curve_mean = curve.groupby(["policy", "t"])["cumulative_expected_reward_chosen"].mean().reset_index()

fig, ax = plt.subplots(figsize=(9, 5))
for pol, g in curve_mean.groupby("policy"):
    ax.plot(g["t"], g["cumulative_expected_reward_chosen"], label=pol)
ax.set_xlabel("Decision step")
ax.set_ylabel("Mean cumulative expected reward")
ax.set_title("Learning curves")
ax.legend(fontsize=8)
fig.tight_layout()
fig.savefig(FIG_DIR / "bandit_learning_curves.png", dpi=200)
plt.show()

action_share = bandit_runs.groupby(["policy", "action"]).size().reset_index(name="n")
action_share["share"] = action_share.groupby("policy")["n"].transform(lambda x: x / x.sum())
display(action_share.pivot(index="policy", columns="action", values="share").fillna(0))
action_share.to_csv(OUTPUT_DIR / "bandit_action_shares.csv", index=False)
```

## 10. Equity and burden diagnostics

```python
# ============================================================
# 10. Group equity and burden evaluation
# ============================================================

def group_eval_for_choices(choices, mats, sim_df, group_col, label):
    idx = np.arange(len(choices))
    p = mats["p_accept"][idx, choices]
    burden = mats["burden"][idx, choices]
    out = []
    groups = sim_df[group_col] if group_col in sim_df.columns else pd.Series(["missing"] * len(sim_df))
    for g, mask_idx in groups.groupby(groups).groups.items():
        mask = np.asarray(list(mask_idx))
        out.append({
            "policy": label,
            "group_col": group_col,
            "group": g,
            "n": len(mask),
            "mean_p_accept": float(np.mean(p[mask])),
            "expected_acceptances": float(np.sum(p[mask])),
            "mean_burden": float(np.mean(burden[mask])),
            "expected_accepted_burden": float(np.sum(p[mask] * burden[mask]) / max(np.sum(p[mask]), 1e-9))
        })
    return out

group_cols = [c for c in ["zero_vehicle_hh", "low_income_hh", "large_apartment_hh", "equity_priority", "home_zone_num"] if c in sim_df.columns]
group_rows = []
for pol in ["HOME_ZONE_ADL", "MIN_BURDEN_VISITED_ADL", "LOWEST_BURDEN_ELIGIBLE", "ORACLE_EXPECTED_REWARD"]:
    for gc in group_cols:
        group_rows.extend(group_eval_for_choices(static_choices[pol], env_mats, sim_df, gc, pol))

group_eval = pd.DataFrame(group_rows)
display(group_eval.head(30))
group_eval.to_csv(OUTPUT_DIR / "static_group_equity_burden_results.csv", index=False)

bandit_group_rows = []
for (pol, run), g in bandit_runs.groupby(["policy", "run"]):
    for gc in group_cols:
        if gc in g.columns:
            gg = g.copy()
        else:
            tmp = sim_df[[gc]].copy().reset_index().rename(columns={"index": "obs_id"})
            gg = g.merge(tmp, on="obs_id", how="left")
        for val, sub in gg.groupby(gc):
            bandit_group_rows.append({
                "policy": pol,
                "run": run,
                "group_col": gc,
                "group": val,
                "n": len(sub),
                "accepted_rate": sub["accepted"].mean(),
                "mean_reward": sub["reward"].mean(),
                "mean_burden": sub["burden"].mean(),
                "mean_burden_if_accepted": np.average(sub["burden"], weights=sub["accepted"]) if sub["accepted"].sum() > 0 else 0.0
            })

bandit_group_eval = pd.DataFrame(bandit_group_rows)
display(bandit_group_eval.groupby(["policy", "group_col", "group"]).mean(numeric_only=True).head(40))
bandit_group_eval.to_csv(OUTPUT_DIR / "bandit_group_equity_burden_results.csv", index=False)
```

## 11. Sensitivity analysis

The recommendation environment is simulated, so a journal paper needs robustness checks. This grid varies opportunity sensitivity, burden sensitivity, equity reward, service value, and burden penalty, then checks whether policy rankings are stable.


```python
# ============================================================
# 11. Sensitivity grid
# ============================================================

def evaluate_static_under_params(params, policies=(
    "HOME_ZONE_ADL", "MIN_BURDEN_VISITED_ADL", "ACTIVITY_PRIORITY_RULE",
    "HIGH_PROPENSITY_MIN_BURDEN", "SCORE_HEURISTIC", "ORACLE_EXPECTED_REWARD"
)):
    at, intercept = add_environment_columns(action_table, params=params, target_mean=float(sim_df[target_col].mean()))
    mats = env_to_matrices(at)
    eq = mats["equity_priority"][:, 0]
    rows = []
    for pol in policies:
        choices = static_policy_choices(pol, mats)
        res = evaluate_choices(choices, mats, eq, pol)
        res["env_intercept"] = intercept
        for k, v in params.items():
            res[f"param_{k}"] = v
        rows.append(res)
    return rows

if RUN_SENSITIVITY:
    sensitivity_rows = []
    for beta_opp in [0.50, 0.90, 1.30]:
        for beta_burden in [0.50, 0.85, 1.20]:
            for equity_bonus in [0.00, 0.18, 0.35]:
                for burden_penalty in [0.12, 0.22, 0.35]:
                    params = dict(ENV_PARAMS_DEFAULT)
                    params.update({
                        "beta_opp": beta_opp,
                        "beta_burden": beta_burden,
                        "equity_bonus": equity_bonus,
                        "burden_penalty": burden_penalty,
                    })
                    sensitivity_rows.extend(evaluate_static_under_params(params))
    sens = pd.DataFrame(sensitivity_rows)
    display(sens.head())
    sens.to_csv(OUTPUT_DIR / "sensitivity_static_policy_results.csv", index=False)

    rank = sens.copy()
    param_cols = [c for c in rank.columns if c.startswith("param_")]
    rank["scenario"] = rank[param_cols].astype(str).agg("|".join, axis=1)
    rank["rank"] = rank.groupby("scenario")["mean_expected_reward"].rank(ascending=False, method="min")
    rank_summary = rank.groupby("policy").agg(
        mean_rank=("rank", "mean"),
        share_rank1=("rank", lambda x: float(np.mean(x == 1))),
        mean_reward=("mean_expected_reward", "mean"),
        mean_burden=("expected_accepted_burden", "mean"),
        mean_equity_gap=("equity_service_gap", "mean")
    ).reset_index().sort_values("mean_rank")
    display(rank_summary)
    rank_summary.to_csv(OUTPUT_DIR / "sensitivity_policy_rank_summary.csv", index=False)
```

# Interpretation and next validation steps

The included run is an exploratory simulator study. Before using this work for a scientific or operational claim:

- select models with nested household-grouped validation rather than ranking alternatives on the final holdout;
- calibrate with household-aware folds and keep final evaluation data untouched;
- build opportunity features without held-out leakage and use time-stamped facility-supply data;
- validate that recommendation-time features are available before the decision;
- estimate action response from randomized or logged recommendation data;
- redesign and stress-test the reward so opportunity and burden are not counted through overlapping terms;
- bootstrap at the household or person level; and
- evaluate fairness as a multi-metric diagnostic, not from a single assumed equity bonus.

Until then, use the outputs to inspect code behavior, assumptions, and sensitivity--not to infer causal impact or real-world policy superiority.
